```python
import zipfile
import os

# Create project directories
RAW_DIR = os.path.join(PROJECT_ROOT, "data", "raw")
os.makedirs(RAW_DIR, exist_ok=True)

# Extract the verified dataset
with zipfile.ZipFile(DATASET_ZIP, 'r') as zip_ref:
    zip_ref.extractall(RAW_DIR)

# Verify extracted files
print("📂 Raw data directory:")
for item in os.listdir(RAW_DIR):
    file_path = os.path.join(RAW_DIR, item)
    print(f"   └── {item} ({os.path.getsize(file_path) / (1024**2):.2f} MB)")

print("\n✅ Dataset extracted successfully.")
```

```python
import pandas as pd
import os

FAKE_PATH = os.path.join(RAW_DIR, "Fake.csv")
TRUE_PATH = os.path.join(RAW_DIR, "True.csv")

# Load datasets
fake_df = pd.read_csv(FAKE_PATH)
true_df = pd.read_csv(TRUE_PATH)

print("📊 Fake News Dataset")
print("Shape:", fake_df.shape)
print("\nColumns:")
print(fake_df.columns.tolist())

print("\n" + "="*60)

print("\n📊 Real News Dataset")
print("Shape:", true_df.shape)
print("\nColumns:")
print(true_df.columns.tolist())

print("\n" + "="*60)

print("\n🔍 Fake Dataset Preview:")
display(fake_df.head())

print("\n🔍 Real Dataset Preview:")
display(true_df.head())
```

```python
print("🔍 FAKE NEWS DATASET")
print("=" * 60)

print("\nMissing values:")
print(fake_df.isnull().sum())

print("\nDuplicate rows:", fake_df.duplicated().sum())


print("\n\n🔍 REAL NEWS DATASET")
print("=" * 60)

print("\nMissing values:")
print(true_df.isnull().sum())

print("\nDuplicate rows:", true_df.duplicated().sum())


print("\n\n📊 SUMMARY")
print("=" * 60)

print(f"Fake rows : {len(fake_df):,}")
print(f"Real rows : {len(true_df):,}")
print(f"Total rows: {len(fake_df) + len(true_df):,}")
```

```python
# Create labels
fake_df["label"] = 0
true_df["label"] = 1

# Combine both datasets
news_df = pd.concat(
    [fake_df, true_df],
    ignore_index=True
)

print("📊 Combined Dataset")
print("=" * 60)

print("Shape:", news_df.shape)

print("\nColumns:")
print(news_df.columns.tolist())

print("\nLabel distribution:")
print(news_df["label"].value_counts())

print("\nLabel percentage:")
print(news_df["label"].value_counts(normalize=True).mul(100).round(2))

print("\n🔹 Label Mapping:")
print("0 → Fake News")
print("1 → Real News")

print("\n🔍 Combined Dataset Preview:")
display(news_df.head())
```

```python
print("📊 SUBJECT DISTRIBUTION")
print("=" * 60)

print("\nOverall subject counts:")
print(news_df["subject"].value_counts())

print("\n\n📊 Subject distribution by label:")
subject_label = pd.crosstab(
    news_df["subject"],
    news_df["label"]
)

subject_label.columns = ["Fake", "Real"]

print(subject_label)

print("\n\n📈 Subject distribution (%):")
subject_percentage = pd.crosstab(
    news_df["subject"],
    news_df["label"],
    normalize="index"
).mul(100).round(2)

subject_percentage.columns = ["Fake %", "Real %"]

display(subject_percentage)
```

```python
# Keep only fields required for the NLP task
news_df = news_df[["title", "text", "label"]].copy()

# Combine title and article text
news_df["content"] = (
    news_df["title"].fillna("") + " " +
    news_df["text"].fillna("")
).str.strip()

# Remove rows with empty content
news_df = news_df[news_df["content"].str.len() > 0].copy()

# Reset index
news_df.reset_index(drop=True, inplace=True)

print("📊 FINAL NLP DATASET")
print("=" * 60)

print("Shape:", news_df.shape)

print("\nColumns:")
print(news_df.columns.tolist())

print("\nLabel distribution:")
print(news_df["label"].value_counts())

print("\nMissing values:")
print(news_df.isnull().sum())

print("\nEmpty content rows:", (news_df["content"].str.len() == 0).sum())

print("\n🔒 Leakage-prone columns removed:")
print("   ❌ subject")
print("   ❌ date")

print("\n✅ Final text feature: title + article text")
```

```python
print("🔍 DUPLICATE ANALYSIS")
print("=" * 60)

# Exact duplicate content
duplicate_content = news_df["content"].duplicated().sum()

# Duplicate content across the same/different labels
duplicate_groups = (
    news_df.groupby("content")["label"]
    .nunique()
)

conflicting_duplicates = (duplicate_groups > 1).sum()

print(f"Exact duplicate articles : {duplicate_content:,}")
print(f"Duplicate groups         : {(duplicate_groups > 1).sum():,}")
print(f"Conflicting-label groups : {conflicting_duplicates:,}")

print("\n📊 Duplicate content by label:")
print(
    news_df.groupby("label")["content"]
    .apply(lambda x: x.duplicated().sum())
)

print("\n🔎 Label mapping:")
print("0 → Fake News")
print("1 → Real News")
```

```python
before_count = len(news_df)

# Remove exact duplicate articles
news_df = news_df.drop_duplicates(
    subset=["content"],
    keep="first"
).reset_index(drop=True)

after_count = len(news_df)
removed_count = before_count - after_count

print("🧹 DUPLICATE REMOVAL")
print("=" * 60)

print(f"Before removing duplicates : {before_count:,}")
print(f"Duplicates removed         : {removed_count:,}")
print(f"After removing duplicates  : {after_count:,}")

print("\n📊 Label distribution after deduplication:")
print(news_df["label"].value_counts())

print("\n📈 Label percentage:")
print(
    news_df["label"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print("\n🔍 Remaining duplicate content:")
print(news_df["content"].duplicated().sum())

print("\n✅ Deduplication completed.")
```

```python
# Calculate text length statistics
news_df["word_count"] = news_df["content"].str.split().str.len()
news_df["char_count"] = news_df["content"].str.len()

print("📊 ARTICLE LENGTH ANALYSIS")
print("=" * 60)

print("\nOverall word-count statistics:")
print(news_df["word_count"].describe().round(2))

print("\nOverall character-count statistics:")
print(news_df["char_count"].describe().round(2))

print("\n📈 Word count by label:")
display(
    news_df.groupby("label")["word_count"]
    .describe()
    .round(2)
)

print("\n📈 Character count by label:")
display(
    news_df.groupby("label")["char_count"]
    .describe()
    .round(2)
)

print("\n🔎 Label mapping:")
print("0 → Fake News")
print("1 → Real News")
```

```python
print("🔍 ARTICLE LENGTH QUALITY CHECK")
print("=" * 60)

# Very short articles
short_articles = news_df[news_df["word_count"] < 20]

print(f"\nArticles with < 20 words: {len(short_articles):,}")

print("\nLabel distribution among very short articles:")
print(short_articles["label"].value_counts())

# Long articles
long_articles = news_df[news_df["word_count"] > 2000]

print(f"\nArticles with > 2,000 words: {len(long_articles):,}")

print("\nLabel distribution among long articles:")
print(long_articles["label"].value_counts())

# Percentiles for detecting extreme lengths
print("\n📈 Word-count percentiles:")
for p in [90, 95, 99, 99.5]:
    print(f"{p}th percentile: {news_df['word_count'].quantile(p/100):.0f} words")

print("\n📌 Note:")
print("We will NOT remove length outliers blindly.")
print("They will be handled during preprocessing/modeling if required.")
```

```python
# Create processed data directory
PROCESSED_DIR = os.path.join(PROJECT_ROOT, "data", "processed")
os.makedirs(PROCESSED_DIR, exist_ok=True)

# Save current cleaned dataset
CLEAN_DATA_PATH = os.path.join(
    PROCESSED_DIR,
    "news_cleaned.csv"
)

news_df.to_csv(CLEAN_DATA_PATH, index=False)

print("💾 CLEANED DATASET SAVED")
print("=" * 60)

print("File:", CLEAN_DATA_PATH)
print(f"Rows saved: {len(news_df):,}")
print(f"Columns saved: {news_df.shape[1]}")

print("\nColumns:")
print(news_df.columns.tolist())

print("\nFile size:")
print(f"{os.path.getsize(CLEAN_DATA_PATH) / (1024**2):.2f} MB")

print("\n✅ Cleaned dataset saved successfully.")
```

```python
from sklearn.model_selection import train_test_split

# First split: 80% train, 20% temporary
train_df, temp_df = train_test_split(
    news_df,
    test_size=0.20,
    stratify=news_df["label"],
    random_state=42
)

# Second split: 10% validation, 10% test
val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    stratify=temp_df["label"],
    random_state=42
)

# Reset indices
train_df = train_df.reset_index(drop=True)
val_df = val_df.reset_index(drop=True)
test_df = test_df.reset_index(drop=True)

print("📊 STRATIFIED DATA SPLIT")
print("=" * 60)

print(f"Total      : {len(news_df):,}")
print(f"Train      : {len(train_df):,}")
print(f"Validation : {len(val_df):,}")
print(f"Test       : {len(test_df):,}")

print("\n📈 Split percentages:")
print(f"Train      : {len(train_df)/len(news_df)*100:.2f}%")
print(f"Validation : {len(val_df)/len(news_df)*100:.2f}%")
print(f"Test       : {len(test_df)/len(news_df)*100:.2f}%")

print("\n🔍 Label distribution:")
print("\nTrain:")
print(train_df["label"].value_counts())

print("\nValidation:")
print(val_df["label"].value_counts())

print("\nTest:")
print(test_df["label"].value_counts())

print("\n📈 Label percentages:")
print("\nTrain:")
print(train_df["label"].value_counts(normalize=True).mul(100).round(2))

print("\nValidation:")
print(val_df["label"].value_counts(normalize=True).mul(100).round(2))

print("\nTest:")
print(test_df["label"].value_counts(normalize=True).mul(100).round(2))

print("\n✅ Stratified 80/10/10 split completed.")
```

```python
print("🔒 SPLIT LEAKAGE CHECK")
print("=" * 60)

train_content = set(train_df["content"])
val_content = set(val_df["content"])
test_content = set(test_df["content"])

train_val_overlap = train_content.intersection(val_content)
train_test_overlap = train_content.intersection(test_content)
val_test_overlap = val_content.intersection(test_content)

print(f"Train ∩ Validation : {len(train_val_overlap):,}")
print(f"Train ∩ Test       : {len(train_test_overlap):,}")
print(f"Validation ∩ Test  : {len(val_test_overlap):,}")

total_overlap = (
    len(train_val_overlap)
    + len(train_test_overlap)
    + len(val_test_overlap)
)

print("\nTotal cross-split overlapping articles:", total_overlap)

if total_overlap == 0:
    print("\n✅ NO DATA LEAKAGE DETECTED")
    print("Train, validation and test sets are completely independent.")
else:
    print("\n❌ DATA LEAKAGE DETECTED")
    print("Overlapping articles found between splits.")
```

```python
# Create split directory
SPLIT_DIR = os.path.join(PROJECT_ROOT, "data", "splits")
os.makedirs(SPLIT_DIR, exist_ok=True)

# File paths
TRAIN_PATH = os.path.join(SPLIT_DIR, "train.csv")
VAL_PATH = os.path.join(SPLIT_DIR, "validation.csv")
TEST_PATH = os.path.join(SPLIT_DIR, "test.csv")

# Save splits
train_df.to_csv(TRAIN_PATH, index=False)
val_df.to_csv(VAL_PATH, index=False)
test_df.to_csv(TEST_PATH, index=False)

print("💾 DATA SPLITS SAVED")
print("=" * 60)

print(f"Train      → {TRAIN_PATH}")
print(f"Validation → {VAL_PATH}")
print(f"Test       → {TEST_PATH}")

print("\n📊 Saved row counts:")
print(f"Train      : {len(train_df):,}")
print(f"Validation : {len(val_df):,}")
print(f"Test       : {len(test_df):,}")

print("\n📁 Split directory contents:")
for file in os.listdir(SPLIT_DIR):
    file_path = os.path.join(SPLIT_DIR, file)
    print(f"   └── {file} ({os.path.getsize(file_path) / (1024**2):.2f} MB)")

print("\n✅ Train/validation/test splits saved successfully.")
```
