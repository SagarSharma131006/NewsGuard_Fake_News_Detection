```python
# ============================================================
# DAY 10 — CELL 6
# RESTORE FINAL 615-FEATURE PIPELINE
# ============================================================

import re, html, unicodedata, string
import numpy as np
import pandas as pd

from gensim.models import Word2Vec
from scipy.sparse import hstack, csr_matrix
from sklearn.feature_selection import SelectKBest, chi2

# ------------------------------------------------------------
# Load Word2Vec
# ------------------------------------------------------------
word2vec_model = Word2Vec.load(
    f"{BASE_DIR}/features/word2vec/word2vec_model.model"
)

assert word2vec_model.vector_size == 100

# ------------------------------------------------------------
# Recreate exact selector
# ------------------------------------------------------------
selector = SelectKBest(chi2, k=500)
selector.fit(train_tfidf, train_labels)

print("Selected TF-IDF features:", selector.get_support().sum())


# ------------------------------------------------------------
# Text cleaning
# ------------------------------------------------------------
def clean_text(text):
    text = html.unescape(str(text))
    text = unicodedata.normalize("NFKC", text)

    text = re.sub(r"<[^>]+>", " ", text)
    text = re.sub(r"https?://\S+|www\.\S+", " url ", text)
    text = re.sub(r"\b[\w\.-]+@[\w\.-]+\.\w+\b", " email ", text)

    text = text.lower()

    text = re.sub(r"[\r\n\t]+", " ", text)

    text = (
        text.replace("–", "-")
            .replace("—", "-")
            .replace("“", '"')
            .replace("”", '"')
            .replace("‘", "'")
            .replace("’", "'")
    )

    text = re.sub(
        r"[^a-z0-9\s\.\,\!\?\:\;\-\(\)\'\"]",
        " ",
        text
    )

    text = re.sub(r"\s+", " ", text).strip()

    return text


# ------------------------------------------------------------
# Auxiliary features — exact 115 features
# ------------------------------------------------------------
def extract_auxiliary_features(text):

    original_text = str(text)
    cleaned = clean_text(original_text)

    words = cleaned.split()

    word_count = len(words)
    char_count = len(cleaned)

    sentence_count = max(
        1,
        len(re.findall(r"[.!?]+", original_text))
    )

    avg_word_length = (
        np.mean([len(w) for w in words])
        if words else 0.0
    )

    avg_sentence_length = (
        word_count / sentence_count
        if sentence_count > 0 else 0.0
    )

    unique_word_ratio = (
        len(set(words)) / word_count
        if word_count > 0 else 0.0
    )

    digit_count = sum(
        char.isdigit()
        for char in original_text
    )

    uppercase_count = sum(
        char.isupper()
        for char in original_text
    )

    punctuation_count = sum(
        char in string.punctuation
        for char in original_text
    )

    # Readability
    import textstat

    flesch_reading_ease = textstat.flesch_reading_ease(
        original_text
    )

    flesch_kincaid_grade = textstat.flesch_kincaid_grade(
        original_text
    )

    gunning_fog = textstat.gunning_fog(
        original_text
    )

    ari = textstat.automated_readability_index(
        original_text
    )

    # Sentiment
    from textblob import TextBlob

    sentiment = TextBlob(original_text).sentiment

    polarity = sentiment.polarity
    subjectivity = sentiment.subjectivity

    # Word2Vec
    tokens = cleaned.split()

    vectors = [
        word2vec_model.wv[word]
        for word in tokens
        if word in word2vec_model.wv
    ]

    if vectors:
        w2v_vector = np.mean(vectors, axis=0)
    else:
        w2v_vector = np.zeros(
            100,
            dtype=np.float32
        )

    # First 15 auxiliary features
    auxiliary = np.array([
        flesch_reading_ease,
        flesch_kincaid_grade,
        gunning_fog,
        ari,
        polarity,
        subjectivity,
        word_count,
        char_count,
        sentence_count,
        avg_word_length,
        avg_sentence_length,
        unique_word_ratio,
        digit_count,
        uppercase_count,
        punctuation_count
    ], dtype=np.float32)

    # Add 100 Word2Vec features
    auxiliary = np.concatenate([
        auxiliary,
        w2v_vector.astype(np.float32)
    ])

    assert auxiliary.shape == (115,)

    return auxiliary


# ------------------------------------------------------------
# Final 615-feature pipeline
# ------------------------------------------------------------
def prepare_final_features(text):

    cleaned = clean_text(text)

    # TF-IDF
    tfidf_features = tfidf_vectorizer.transform(
        [cleaned]
    )

    # Select 500 TF-IDF features
    selected_tfidf = selector.transform(
        tfidf_features
    )

    # 115 auxiliary features
    auxiliary_features = extract_auxiliary_features(
        text
    )

    auxiliary_sparse = csr_matrix(
        auxiliary_features.reshape(1, -1)
    )

    # 500 + 115 = 615
    final_features = hstack([
        selected_tfidf,
        auxiliary_sparse
    ]).tocsr()

    return final_features


# ------------------------------------------------------------
# Verification
# ------------------------------------------------------------
pipeline_test = prepare_final_features(
    "This is a test news article about economic policy."
)

print("Final feature shape:", pipeline_test.shape)

assert pipeline_test.shape == (1, 615)

print("Word2Vec vocabulary:", len(word2vec_model.wv))
print("Word2Vec dimensions:", word2vec_model.vector_size)
print("Final features:", pipeline_test.shape[1])

print("\n✅ DAY 10 — CELL 6 PIPELINE RESTORATION PASSED")
```

```python
# ============================================================
# DAY 10 — CELL 7A
# RESTORE FINAL FEATURE NAMES
# ============================================================

# 500 selected TF-IDF feature names
selected_tfidf_names = (
    np.array(tfidf_vectorizer.get_feature_names_out())
    [selector.get_support()]
    .tolist()
)

# 15 auxiliary feature names
base_auxiliary_names = [
    "Flesch Reading Ease",
    "Flesch-Kincaid Grade",
    "Gunning Fog",
    "ARI",
    "TextBlob Polarity",
    "TextBlob Subjectivity",
    "Word Count",
    "Character Count",
    "Sentence Count",
    "Average Word Length",
    "Average Sentence Length",
    "Unique Word Ratio",
    "Digit Count",
    "Uppercase Count",
    "Punctuation Count"
]

# 100 Word2Vec feature names
w2v_names = [
    f"W2V_{i:03d}"
    for i in range(1, 101)
]

auxiliary_feature_names = (
    base_auxiliary_names + w2v_names
)

feature_names = (
    selected_tfidf_names +
    auxiliary_feature_names
)

print("Selected TF-IDF names:", len(selected_tfidf_names))
print("Auxiliary names:", len(auxiliary_feature_names))
print("Total feature names:", len(feature_names))

assert len(selected_tfidf_names) == 500
assert len(auxiliary_feature_names) == 115
assert len(feature_names) == 615

print("\n✅ DAY 10 — CELL 7A FEATURE NAMES RESTORED")
```

```python
# ============================================================
# DAY 10 — CELL 7
# SHAP + FINAL PREDICTION VERIFICATION
# ============================================================

import json
import numpy as np
import pandas as pd

# ------------------------------------------------------------
# Test article
# ------------------------------------------------------------
sample_text = """
The government announced a new economic policy today.
Officials said the policy will support employment and
increase investment in infrastructure and public services.
"""

# ------------------------------------------------------------
# Prepare final 615 features
# ------------------------------------------------------------
sample_features = prepare_final_features(sample_text)

assert sample_features.shape == (1, 615)

# ------------------------------------------------------------
# Model prediction
# ------------------------------------------------------------
prediction = best_xgb.predict(sample_features)[0]

probabilities = best_xgb.predict_proba(sample_features)[0]

fake_probability = float(probabilities[0])
real_probability = float(probabilities[1])

credibility_score = real_probability * 100

label = "Real" if prediction == 1 else "Fake"

# ------------------------------------------------------------
# SHAP values
# ------------------------------------------------------------
sample_dense = sample_features.toarray()

sample_shap = shap_explainer.shap_values(sample_dense)

if isinstance(sample_shap, list):
    sample_shap = sample_shap[0]

sample_shap = np.asarray(sample_shap)

if sample_shap.ndim == 3:
    sample_shap = sample_shap[:, :, 0]

sample_shap_values = sample_shap[0]

assert len(sample_shap_values) == 615

# ------------------------------------------------------------
# Top SHAP features
# ------------------------------------------------------------
feature_names = selected_tfidf_names + auxiliary_feature_names

assert len(feature_names) == 615

shap_df = pd.DataFrame({
    "feature": feature_names,
    "shap_value": sample_shap_values
})

shap_df["abs_shap"] = shap_df["shap_value"].abs()

top_shap_features = (
    shap_df
    .sort_values("abs_shap", ascending=False)
    .head(10)
)

# ------------------------------------------------------------
# Display
# ------------------------------------------------------------
print("Prediction:", label)
print(f"Fake probability: {fake_probability:.6f}")
print(f"Real probability: {real_probability:.6f}")
print(f"Credibility score: {credibility_score:.2f}/100")

print("\nTop SHAP features:")
display(
    top_shap_features[
        ["feature", "shap_value"]
    ]
)

# ------------------------------------------------------------
# Save verification metadata
# ------------------------------------------------------------
day10_shap_prediction = {
    "status": "COMPLETE",
    "prediction": label,
    "fake_probability": fake_probability,
    "real_probability": real_probability,
    "credibility_score": credibility_score,
    "feature_count": int(sample_features.shape[1]),
    "top_shap_features": [
        {
            "feature": str(row["feature"]),
            "shap_value": float(row["shap_value"])
        }
        for _, row in top_shap_features.iterrows()
    ]
}

with open(
    f"{DAY10_RESULTS_DIR}/day_10_shap_prediction_verification.json",
    "w"
) as f:
    json.dump(
        day10_shap_prediction,
        f,
        indent=2
    )

print("\nSaved:")
print("→ day_10_shap_prediction_verification.json")

print("\n✅ DAY 10 — CELL 7 SHAP + PREDICTION VERIFICATION PASSED")
```
