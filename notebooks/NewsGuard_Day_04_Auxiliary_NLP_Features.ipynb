```python
# ================================================================
# NewsGuard — Day 4
# Cell 3: Word2Vec Setup
# ================================================================

from gensim.models import Word2Vec
import numpy as np

# Tokenize text
train_tokens = [
    text.split()
    for text in X_train_text
]

val_tokens = [
    text.split()
    for text in X_val_text
]

test_tokens = [
    text.split()
    for text in X_test_text
]

print("=" * 64)
print("WORD2VEC DATA PREPARATION")
print("=" * 64)

print("\nTokenized documents:")
print(f"Train      : {len(train_tokens):,}")
print(f"Validation : {len(val_tokens):,}")
print(f"Test       : {len(test_tokens):,}")

print("\nExample tokens:")
print(train_tokens[0][:20])

assert len(train_tokens) == 31282
assert len(val_tokens) == 3910
assert len(test_tokens) == 3911

assert all(len(tokens) > 0 for tokens in train_tokens)
assert all(len(tokens) > 0 for tokens in val_tokens)
assert all(len(tokens) > 0 for tokens in test_tokens)

print("\n" + "=" * 64)
print("WORD2VEC INPUT PREPARATION PASSED")
print("=" * 64)
```

```python
# ================================================================
# NewsGuard — Day 4
# Cell 4: Fast Word2Vec Training
# ================================================================

from gensim.models import Word2Vec

print("=" * 64)
print("TRAINING FAST WORD2VEC")
print("=" * 64)

w2v_model = Word2Vec(
    sentences=train_tokens,
    vector_size=100,
    window=5,
    min_count=2,
    workers=4,
    sg=1,
    epochs=2,
    seed=42,
    compute_loss=False
)

print("\nWord2Vec configuration:")
print(f"Vector size : {w2v_model.vector_size}")
print(f"Window      : {w2v_model.window}")
print(f"Min count   : {w2v_model.min_count}")
print(f"Epochs      : {w2v_model.epochs}")
print(f"Vocabulary  : {len(w2v_model.wv):,}")

assert w2v_model.vector_size == 100
assert len(w2v_model.wv) > 0

print("\n" + "=" * 64)
print("WORD2VEC TRAINING PASSED")
print("Model trained using TRAINING DATA ONLY.")
print("=" * 64)
```

```python
# ================================================================
# NewsGuard — Day 4
# Cell 5: Create Sentence-Level Word2Vec Features
# ================================================================

import numpy as np

def document_vector(tokens, model):
    vectors = [
        model.wv[word]
        for word in tokens
        if word in model.wv
    ]

    if len(vectors) == 0:
        return np.zeros(model.vector_size, dtype=np.float32)

    return np.mean(vectors, axis=0).astype(np.float32)


print("=" * 64)
print("CREATING WORD2VEC SENTENCE VECTORS")
print("=" * 64)

X_train_w2v = np.vstack([
    document_vector(tokens, w2v_model)
    for tokens in train_tokens
])

X_val_w2v = np.vstack([
    document_vector(tokens, w2v_model)
    for tokens in val_tokens
])

X_test_w2v = np.vstack([
    document_vector(tokens, w2v_model)
    for tokens in test_tokens
])

print("\nFeature shapes:")
print(f"Train      : {X_train_w2v.shape}")
print(f"Validation : {X_val_w2v.shape}")
print(f"Test       : {X_test_w2v.shape}")

print("\nData type:")
print(f"Train      : {X_train_w2v.dtype}")
print(f"Validation : {X_val_w2v.dtype}")
print(f"Test       : {X_test_w2v.dtype}")

# Verify dimensions
assert X_train_w2v.shape == (31282, 100)
assert X_val_w2v.shape == (3910, 100)
assert X_test_w2v.shape == (3911, 100)

assert X_train_w2v.dtype == np.float32
assert X_val_w2v.dtype == np.float32
assert X_test_w2v.dtype == np.float32

# Check for invalid values
assert np.isfinite(X_train_w2v).all()
assert np.isfinite(X_val_w2v).all()
assert np.isfinite(X_test_w2v).all()

print("\n" + "=" * 64)
print("WORD2VEC SENTENCE VECTOR CREATION PASSED")
print("=" * 64)
```

```python
# ================================================================
# CORRECTED LINGUISTIC / ARTICLE-LEVEL FEATURE EXTRACTION
# ================================================================

def linguistic_features(clean_text, original_text):
    clean_text = str(clean_text)
    original_text = str(original_text)

    words = clean_text.split()
    word_count = len(words)

    char_count = len(clean_text)

    sentence_count = max(
        1,
        clean_text.count(".") +
        clean_text.count("!") +
        clean_text.count("?")
    )

    avg_word_length = (
        sum(len(word) for word in words) / word_count
        if word_count > 0 else 0.0
    )

    avg_sentence_length = word_count / sentence_count

    unique_word_ratio = (
        len(set(words)) / word_count
        if word_count > 0 else 0.0
    )

    digit_count = sum(char.isdigit() for char in original_text)

    # IMPORTANT:
    # Count uppercase letters from ORIGINAL content
    uppercase_count = sum(char.isupper() for char in original_text)

    punctuation_count = sum(
        char in ".,!?;:'\"-()[]{}"
        for char in original_text
    )

    return [
        word_count,
        char_count,
        sentence_count,
        avg_word_length,
        avg_sentence_length,
        unique_word_ratio,
        digit_count,
        uppercase_count,
        punctuation_count
    ]


X_train_linguistic = np.array(
    [
        linguistic_features(clean_text, original_text)
        for clean_text, original_text
        in zip(train_df["clean_content"], train_df["content"])
    ],
    dtype=np.float32
)

X_val_linguistic = np.array(
    [
        linguistic_features(clean_text, original_text)
        for clean_text, original_text
        in zip(val_df["clean_content"], val_df["content"])
    ],
    dtype=np.float32
)

X_test_linguistic = np.array(
    [
        linguistic_features(clean_text, original_text)
        for clean_text, original_text
        in zip(test_df["clean_content"], test_df["content"])
    ],
    dtype=np.float32
)

print("=" * 64)
print("CORRECTED LINGUISTIC FEATURE EXTRACTION")
print("=" * 64)

print("\nFeature shapes:")
print(f"Train      : {X_train_linguistic.shape}")
print(f"Validation : {X_val_linguistic.shape}")
print(f"Test       : {X_test_linguistic.shape}")

print("\nFeature names:")
feature_names = [
    "Word Count",
    "Character Count",
    "Sentence Count",
    "Average Word Length",
    "Average Sentence Length",
    "Unique Word Ratio",
    "Digit Count",
    "Uppercase Count",
    "Punctuation Count"
]

for i, name in enumerate(feature_names, 1):
    print(f"{i}. {name}")

print("\nTRAIN statistics:")
print("-" * 64)

for i, name in enumerate(feature_names):
    print(
        f"{name:28} "
        f"mean={X_train_linguistic[:, i].mean():.4f} | "
        f"std={X_train_linguistic[:, i].std():.4f}"
    )

assert X_train_linguistic.shape == (31282, 9)
assert X_val_linguistic.shape == (3910, 9)
assert X_test_linguistic.shape == (3911, 9)

assert np.isfinite(X_train_linguistic).all()
assert np.isfinite(X_val_linguistic).all()
assert np.isfinite(X_test_linguistic).all()

assert np.all(X_train_linguistic[:, 0] >= 0)
assert np.all(X_train_linguistic[:, 1] >= 0)
assert np.all(X_train_linguistic[:, 2] >= 1)
assert np.all(X_train_linguistic[:, 5] >= 0)
assert np.all(X_train_linguistic[:, 5] <= 1)
assert np.all(X_train_linguistic[:, 7] >= 0)

print("\nUPPERCASE FEATURE CORRECTION VERIFIED")
print("\nLINGUISTIC FEATURE EXTRACTION PASSED")
```
